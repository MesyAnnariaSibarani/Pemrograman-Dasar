```python
# Nama    : Mesy Annaria Sibarani
# NIM     : 4232501018
# Tanggal : 23 September 2026
# Tujuan  : Menghitung dan menampilkan hasil penjumlahan, pengurangan, perkalian, pembagian biasa, pembagian bulat, dan sisa bagi (modulus).

# B.1 Soal Umum

# Jawaban 1
Bilangan_pertama = 200000
Bilangan_kedua   = 20000

# Penjumlahan
Total_bilangan = (Bilangan_pertama + Bilangan_kedua)
print(Total_bilangan)

# Pengurangan
Total_bilangan = (Bilangan_pertama - Bilangan_kedua)
print(Total_bilangan)

# Perkalian
Total_bilangan = (Bilangan_pertama * Bilangan_kedua)
print(Total_bilangan)

# Pembagian
Total_bilangan = (Bilangan_pertama / Bilangan_kedua)
print(Total_bilangan)

# Pembagian Bulat
Total_bilangan = (Bilangan_pertama // Bilangan_kedua)
print(Total_bilangan)

# Modulus
Total_bilangan = (Bilangan_pertama % Bilangan_kedua)
print(Total_bilangan)
```

```python
# Nama    : Mesy Annaria Sibarani
# NIM     : 4232501018
# Tanggal : 23 September 2026
# Tujuan  : Untuk mengevaluasi secara otomatis apakah seorang siswa lulus atau tidak berdasarkan nilai ujian yang dimasukkan, lalu menampilkan hasilnya dalam bentuk nilai logika (True atau False).

# Jawaban 2
# Meminta pengguna memasukkan nilai ujian (0-100)
nilai = float(input("Masukkan nilai ujian (0-100): "))

# Menggunakan operator relasional (>=) untuk memeriksa kelulusan
nilai_ini_termasuk_kategori_lulus  = nilai >= 60

# Menampilkan hasil True atau False
print("Nilai Termasuk Kategori lulus (≥ 60):", nilai_ini_termasuk_kategori_lulus)
```

```python
# Nama    : Mesy Annaria Sibarani
# NIM     : 4232501018
# Tanggal : 23 September 2026
# Tujuan  : Melakukan validasi logis terhadap kelayakan seseorang untuk mengemudi sendiri berdasarkan hukum dan kondisi keselamatan.

# Jawaban 3
# Meminta input umur dari pengguna
umur = int(input("Masukkan umur Anda: "))

# Meminta input status SIM (True/False)
input_sim = input("Apakah Anda punya SIM? (True/False): ")
punya_sim = input_sim.strip().lower() == 'true'

# Meminta input tambahan untuk contoh operator precedence (izin khusus/darurat)
input_KTP = input("Apakah punya KTP? (True/False): ")
KTP = input_KTP.strip().lower() == 'true'

# EKSPRESI DENGAN TANDA KURUNG (Operator Precedence)
# Tanpa tanda kurung: umur >= 17 or KTP and punya_sim
# (karena 'and' lebih kuat dari 'or', program akan mengira: umur >= 17 or (KTP and punya_sim))
# Dengan tanda kurung di bawah, kita memaksa agar (umur >= 17 or KTP) dievaluasi terlebih dahulu.
boleh_mengemudi = (umur >= 17 or KTP) and punya_sim

# Menampilkan hasil keputusan
if boleh_mengemudi:
    print("Hasil: Anda boleh mengemudi sendiri.")
else:
    print("Hasil: Anda TIDAK boleh mengemudi sendiri.")
```

```python
# Nama    : Mesy Annaria Sibarani
# NIM     : 4232501018
# Tanggal : 23 September 2026
# Tujuan  : Untuk menghitung persentase efisiensi termal dari sebuah unit pembangkit listrik berdasarkan nilai energi input dan energi output yang dimasukkan oleh pengguna.

# B.2 Soal Konteks Energi

# Jawaban 4
# Meminta input dari pengguna
energi_input = float(input("Masukkan nilai energi input (dalam MJ): "))
energi_output = float(input("Masukkan nilai energi output (dalam MJ): "))

# Menghitung efisiensi termal dengan operator precedence yang benar
# Tanda kurung memastikan pembagian dilakukan terlebih dahulu sebelum dikalikan 100
efisiensi = (energi_output / energi_input) * 100

# Menampilkan hasil dengan f-string dan dibulatkan 2 angka desimal
print(f"Efisiensi termal unit pembangkit adalah: {efisiensi:.2f}%")
```

```python
# Nama    : Mesy Annaria Sibarani
# NIM     : 4232501018
# Tanggal : 23 September 2026
# Tujuan  : Untuk memantau dan mengevaluasi status keamanan operasi boiler pada suatu unit pembangkit listrik.

# Jawaban 5
# Meminta pengguna memasukkan nilai suhu boiler dan beban operasi
suhu_boiler = float(input("Masukkan suhu boiler (°C): "))
beban_operasi = float(input("Masukkan beban operasi (%): "))

# Menghitung kondisi perlu maintenance menggunakan operator relasional dan logika
perlu_maintenance = (suhu_boiler > 550) or (beban_operasi > 95)

# Menampilkan hasil dalam satu kalimat naratif menggunakan f-string
print(f"Berdasarkan suhu boiler {suhu_boiler}°C dan beban operasi {beban_operasi}%, status unit perlu maintenance adalah {perlu_maintenance}.")
```
