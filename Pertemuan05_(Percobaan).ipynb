```python
#Definisikan_Variabel

harga_gorengan = 2500
jumlah_gorengan = 3
harga_esteh = 4000
jumlah_esteh = 2
uang = 20000

harga_gorengan = int(input("Masukkan Harga Gorengan : "))
jumlah_gorengan = int(input("Masukkan Jumlah Gorengan : "))
harga_esteh = int(input("Masukkan Harga Esteh : "))
jumlah_esteh = int(input("Masukkan Jumlah Esteh : "))
uang = int(input("Masukkan Uang : "))

#Operasikan
#total Belanja
total = harga_gorengan * jumlah_gorengan + harga_esteh * jumlah_esteh
print("Total:", total)

#Uang Kembalian
kembalian = uang - total
print("Kembalian:", kembalian)

#status uang cukup
cukup = uang >= total
print("Cukup:", cukup)
```

```python
#Todo
#1. jam penuh = total menit dibagi 60 tanpa pecahan (//)
#2. menit sisa = sisa pembagian total menit dengan 60 (%)
#3. jam desimal = total menit dibagi 60 dengan pecahan (/)
#4. cek rentang dengan > dan < yang digabung and.

#Durasi_dalam_menit
total_menit = 275
jam_penuh = total_menit // 60
sisa_menit = total_menit % 60
jam_desimal = total_menit / 60
dalam_rentang = total_menit >= 180 and total_menit <= 300

print("Durasi :", jam_penuh, "jam", sisa_menit, "menit")
print("jam desimal :", round(jam_desimal,2))
print("3 s.d. 5 jam?:", dalam_rentang)
```
