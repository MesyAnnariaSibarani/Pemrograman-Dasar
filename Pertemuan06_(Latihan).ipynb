```python
# 1. Buat program yang meminta sebuah angka, lalu tampilkan "Kelipatan 5" jika angka tersebut habis dibagi 5.

# Meminta input angka dari pengguna
angka = int(input("Masukkan sebuah angka: "))

# Memeriksa apakah angka habis dibagi 5
if angka % 5 == 0:
    print("Kelipatan 5")
```

```python
# 2. Menetukan Tahun Kabisat dengan menggunakan kondisi
  # a. Penjelasan:
    # a.1. Syarat tahun kabisat:
      # Habis dibagi 400 -> Kabisat.
      # Habis dibagi 4 dan tidak habis dibagi 100 -> Kabisat.

tahun = int(input("Masukkan Tahun: "))

# Logika kondisi sesuai syarat di atas
if (tahun % 400 == 0) or (tahun % 4 == 0 and tahun % 100 != 0):
    print(f"{tahun} adalah Tahun Kabisat")
else:
    print(f"{tahun} Bukan Tahun Kabisat")
```
