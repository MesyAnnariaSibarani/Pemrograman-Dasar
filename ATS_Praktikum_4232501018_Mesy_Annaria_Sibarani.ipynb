```python
# Soal 2. Energi Harian dan Capacity Factor PLTD

# Menyimpan nilai dalam variabel
daya_kw = 450
jam = 8
energi_aktual = 7200
kapasitas_kw = 400

# a. Menghitung energi harian dan menampilkan dengan f-string
energi = daya_kw * jam
print(f"Energi harian = {energi} kWh")

# b. Menghitung capacity factor dan menampilkan dengan 2 angka desimal
capacity_factor = (energi_aktual / (kapasitas_kw * 24)) * 100
print(f"Capacity factor = {capacity_factor:.2f}%")


#PENJELASAN
# • daya_kw = 450: Menyimpan nilai daya output pembangkit sebesar 450 kW ke dalam variabel.
# • jam = 8: Menyimpan durasi waktu operasi selama 8 jam.
# • energi_aktual = 7200: Menyimpan total produksi energi aktual (7200 kWh) untuk perhitungan faktor kapasitas.
# • kapasitas_kw = 400: Menyimpan nilai kapasitas maksimum terpasang dari pembangkit (400 kW).
# • energi = daya_kw * jam: Menghitung energi harian dengan mengalikan daya operasi dengan jumlah jam kerja.
# • print(f"Energi harian = {energi} kWh"): Menampilkan hasil perhitungan energi harian menggunakan format f-string.
# • capacity_factor = (energi_aktual / (kapasitas_kw * 24)) * 100: Menghitung rasio energi aktual terhadap kapasitas maksimum teoritis selama 24 jam penuh dalam bentuk persentase.
# • print(f"Capacity factor = {capacity_factor:.2f}%"): Menampilkan hasil faktor kapasitas dengan pembatasan 2 angka di belakang desimal (:.2f).
```

```python
# Soal 6. Status Operasi Boiler

# 1. Membaca input suhu dan tekanan dari user dan mengubahnya menjadi tipe float
suhu = float(input("Masukkan suhu uap (°C): "))
tekanan = float(input("Masukkan tekanan (bar): "))

# 2. Pemeriksaan kondisi dengan struktur if-elif-else
# Kondisi TRIP diperiksa pertama karena memiliki prioritas tertinggi (kritis)
if suhu >= 550 or tekanan >= 14:
    print("STATUS: TRIP")

# Kondisi WARNING diperiksa jika kondisi TRIP tidak terpenuhi
elif suhu > 520 and tekanan > 12:
    print("STATUS: WARNING")

# Jika semua kondisi di atas tidak terpenuhi, maka status aman/normal
else:
    print("STATUS: NORMAL")


# PENJELASAN
# 1. Menerima Input: Program meminta pengguna untuk memasukkan dua nilai, yaitu nilai suhu uap (dalam satuan °C) dan nilai tekanan (dalam satuan bar).
# 2. Konversi Tipe Data: Input teks dari pengguna langsung diubah menjadi angka desimal (float) agar bisa digunakan dalam perhitungan atau perbandingan matematis.
# 3. Evaluasi Kondisi (Logika Percabangan): Program memeriksa tingkat keamanan boiler menggunakan struktur if-elif-else dengan aturan prioritas dari yang paling kritis:
	#• Status TRIP (Kritis): Jika suhu mencapai 550°C atau lebih, ATAU tekanan mencapai 14 bar atau lebih, sistem langsung menyatakan status TRIP (mati mendadak demi keselamatan).
	#• Status WARNING (Peringatan): Jika tidak memenuhi kondisi TRIP, program mengecek apakah suhu di atas 520°C DAN tekanan di atas 12 bar secara bersamaan. Jika ya, sistem memberikan status WARNING.
  #• Status NORMAL (Aman): Jika seluruh kondisi bahaya di atas tidak terpenuhi, maka boiler dinyatakan dalam status NORMAL.

# • suhu = float(input("Masukkan suhu uap (°C): ")) Mengambil input teks suhu dari pengguna melalui fungsi input(), mengubahnya menjadi bilangan desimal dengan float(), lalu menyimpannya ke dalam variabel suhu.
# • tekanan = float(input("Masukkan tekanan (bar): ")) Mengambil input teks tekanan dari pengguna, mengubahnya menjadi bilangan desimal dengan float(), lalu menyimpannya ke dalam variabel tekanan.
# • if suhu >= 550 or tekanan >= 14: Evaluasi kondisi pertama (prioritas tertinggi). Menggunakan operator logika or (atau), artinya jika salah satu atau kedua kondisi terpenuhi (suhu \(\ge \) 550 atau tekanan \(\ge \) 14), maka program akan menjalankan baris kode di bawahnya.
# • print("STATUS: TRIP") Mencetak teks "STATUS: TRIP" ke layar jika kondisi pada blok if pertama terpenuhi.
# • elif suhu > 520 and tekanan > 12: Evaluasi kondisi kedua apabila kondisi pertama tidak terpenuhi. Menggunakan operator logika and (dan), artinya kondisi ini hanya terpenuhi jika kedua syarat terjadi bersamaan (suhu > 520 sekaligus tekanan > 12).
# • print("STATUS: WARNING") Mencetak teks "STATUS: WARNING" ke layar jika kondisi pada blok elif terpenuhi.
# • else: Merupakan jalur alternatif terakhir jika seluruh kondisi penyaringan di atas (if dan elif) bernilai salah (False).
# • print("STATUS: NORMAL") Mencetak teks "STATUS: NORMAL" ke layar sebagai penanda bahwa boiler bekerja pada batas parameter yang aman.
```

```python
# Soal 8. Pengosongan Tangki Bahan Bakar

# Inisialisasi volume awal tangki sebesar 100 liter
volume = 100

# Perulangan berjalan selama volume tangki lebih besar dari 0
while volume > 0:
    # Kurangi volume tangki sebanyak 20 liter di setiap iterasi
    volume -= 20

    # Cetak volume saat ini setelah dikurangi
    print(f"Volume: {volume} liter")

    # Jika volume tepat mencapai 20 liter, cetak pesan peringatan
    if volume == 20:
        print("WARNING: Hampir kosong!")

# Pesan yang dicetak setelah perulangan while selesai (tangki sudah 0 liter)
print("TANGKI KOSONG - Pengosongan berhenti.")


# PENJELASAN
# Program ini berfungsi untuk mensimulasikan proses pengurasan atau pengosongan tangki bahan bakar secara bertahap dari kondisi penuh hingga habis total.
# 1. Inisialisasi Awal: Tangki dimulai dengan kapasitas penuh sebesar 100 liter.
# 2. Perulangan Pengosongan: Program menggunakan struktur perulangan while untuk terus mengurangi isi tangki secara berkala sebanyak 20 liter di setiap tahapnya selama bahan bakar masih ada (volume > 0). Sisa volume pada setiap tahap akan langsung ditampilkan di layar.
# 3. Peringatan Batas Minimum: Di dalam proses pengurangan tersebut, sistem secara cerdas memantau volume tangki. Ketika bahan bakar menyusut hingga tepat menyisakan 20 liter, program akan memicu pesan peringatan (WARNING).
# 4. Penghentian Otomatis: Begitu bahan bakar habis menyentuh angka 0 liter, kondisi perulangan while menjadi tidak terpenuhi sehingga proses pengurangan otomatis berhenti, lalu program menutup simulasi dengan menyatakan bahwa tangki telah kosong.

# • volume = 100 Berfungsi untuk membuat variabel baru bernama volume dan menginisialisasi atau mengisi nilai awalnya sebesar 100 (merepresentasikan 100 liter bahan bakar awal).
# • while volume > 0: Merupakan perintah perulangan (looping). Baris kode di bawahnya akan terus dijalankan berulang kali selama nilai di dalam variabel volume masih lebih besar dari 0.
# • volume -= 20 Berfungsi untuk mengurangi nilai di dalam variabel volume sebanyak 20 pada setiap putaran operasi. Kode ini merupakan bentuk singkat dari perintah volume = volume - 20.
# • print(f"Volume: {volume} liter") Berfungsi untuk mencetak dan menampilkan sisa volume tangki bahan bakar yang paling mutakhir ke layar setelah melalui proses pengurangan, memanfaatkan format f-string agar variabel dapat disisipkan langsung di dalam teks.
# • if volume == 20: Merupakan struktur percabangan logika atau kondisional untuk memeriksa apakah sisa bahan bakar di dalam variabel volume saat ini nilainya tepat sama dengan 20.
# • print("WARNING: Hampir kosong!") Berfungsi untuk mencetak teks peringatan ke layar. Baris ini hanya akan dieksekusi apabila kondisi pada perintah if di atasnya terpenuhi (ketika volume bernilai tepat 20).
# print("TANGKI KOSONG - Pengosongan berhenti.") Berstatus di luar blok perulangan while. Perintah ini hanya akan dijalankan satu kali di akhir program setelah nilai volume mencapai 0 (atau kurang) dan perulangan while telah sepenuhnya selesai/berhenti.
```

```python
# Soal 10. Pembacaan Suhu Bearing dengan Sensor Rusak

# Data suhu bearing lima titik pengukuran
data = ["62.5", "58.0", "81.3", "N/A", "66.0"]

# Inisialisasi penghitung jumlah data valid
jumlah_data_valid = 0

# Perulangan untuk memeriksa setiap data (nomor titik dimulai dari 1)
for n, nilai_teks in enumerate(data, start=1):
    try:
        # a. Mengubah data menjadi float
        nilai_float = float(nilai_teks)

        # c. Menghitung jumlah data valid yang berhasil diperiksa
        jumlah_data_valid += 1

        # b. Jika nilai melebihi 70.0, cetak peringatan
        if nilai_float > 70.0:
            print(f"Warning: Titik {n} suhu bearing tinggi!")

    except ValueError:
        # a. Jika gagal konversi, cetak pesan error lalu hentikan loop
        print(f"Titik {n}: data tidak valid - pemeriksaan dihentikan")
        break

# c. Tampilkan jumlah data valid setelah perulangan selesai
print(f"Jumlah data valid: {jumlah_data_valid}")

# PENJELASAN
# 1. Inisialisasi Data: Program menyiapkan sebuah daftar (list) berisi 5 data suhu dalam bentuk teks, di mana salah satunya bernilai "N/A" (merepresentasikan sensor rusak). Variabel hitung juga disiapkan untuk mencatat jumlah data yang sukses diproses.
# 2. Perulangan dan Penomoran: Program melakukan looping untuk membaca setiap data suhu sekaligus memberikan penomoran titik yang dimulai dari angka 1.
# 3. Validasi dan Konversi (Blok try): Setiap data teks dicoba untuk diubah menjadi angka desimal (float).
	#• Jika berhasil, jumlah data valid bertambah 1, lalu program mengecek apakah suhunya melebihi 70.0°C untuk memberikan peringatan.
# 4. Penanganan Eror (Blok except): Jika konversi gagal (seperti saat bertemu teks "N/A"), program akan memicu ValueError. Program langsung mencetak pesan eror dan menghentikan seluruh proses pemeriksaan menggunakan perintah break.
# 5. Output Akhir: Setelah perulangan selesai atau dihentikan, program menampilkan total data valid yang berhasil diperiksa.

# • data = ["62.5", "58.0", "81.3", "N/A", "66.0"] Membuat variabel data bertipe list untuk menyimpan string teks hasil pengukuran suhu dari lima titik bearing.
# • jumlah_data_valid = 0 Membuat variabel penghitung (counter) dengan nilai awal 0 untuk menghitung berapa banyak data yang berhasil dikonversi dan diperiksa.
# • for n, nilai_teks in enumerate(data, start=1): Melakukan perulangan untuk mengekstrak isi list. Fungsi enumerate(..., start=1) digunakan agar variabel n menyimpan nomor urut titik (dimulai dari 1) dan nilai_teks menyimpan nilai suhunya.
# • try: Memulai blok pengujian untuk kode yang berpotensi menimbulkan eror (dalam hal ini, kegagalan saat mengubah teks menjadi angka).
# • nilai_float = float(nilai_teks) Mengubah tipe data nilai_teks yang awalnya berupa string/teks menjadi bilangan desimal (float) agar bisa dibandingkan secara matematis.
# • jumlah_data_valid += 1 Menambahkan nilai variabel jumlah_data_valid sebanyak 1 poin, menandakan bahwa satu data telah sukses melewati proses konversi tanpa eror.
# • if nilai_float > 70.0: Struktur logika percabangan untuk memeriksa apakah nilai suhu yang telah dikonversi tersebut bernilai lebih besar dari 70.0.
# • print(f"Warning: Titik {n} suhu bearing tinggi!") Mencetak pesan peringatan ke layar yang menyebutkan nomor titik n jika kondisi suhu di atas 70.0 terpenuhi.
# • except ValueError: Blok yang akan langsung dieksekusi apabila kode di dalam blok try mengalami eror spesifik berupa ValueError (gagal melakukan konversi tipe data).
# • print(f"Titik {n}: data tidak valid - pemeriksaan dihentikan") Mencetak pemberitahuan ke layar bahwa pada nomor titik tertentu terdapat data yang rusak sehingga pemeriksaan harus disudahi.
# • break: Perintah untuk keluar secara paksa dari perulangan for, sehingga sisa data di dalam list (seperti data kelima "66.0") tidak akan diproses lagi.
# • print(f"Jumlah data valid: {jumlah_data_valid}") Baris terakhir di luar loop yang berfungsi menampilkan ringkasan total data suhu yang berhasil diverifikasi dengan aman.
```

```python
# Challenge: Bonus Nilai

"""
DESKRIPSI MASALAH DAN CARA KERJA PROGRAM:
Masalah:
Sebuah fasilitas pembangkit listrik mikro/gardu energi perlu mencatat beban listrik
harian (dalam kW) untuk mendeteksi apakah terjadi pemakaian normal, tinggi, atau
bahkan anomali beban berlebih (overload) yang berpotensi merusak sistem grid.

Cara Kerja Program:
1. Program meminta pengguna memasukkan data beban listrik satu per satu secara interaktif.
2. Menggunakan blok `try-except` untuk memastikan input yang dimasukkan berupa angka valid (float).
3. Data disimpan ke dalam sebuah `list`. Input dihentikan jika pengguna mengetik 'selesai'.
4. Setelah data terkumpul, program melakukan perulangan `for` untuk memeriksa setiap data beban:
   - Jika beban > 500 kW -> Kategori: ANOMALI / OVERLOAD
   - Jika beban > 350 kW -> Kategori: WARNING / BEBAN TINGGI
   - Selain itu          -> Kategori: NORMAL
5. Di akhir, program menghitung rata-rata beban harian dan menampilkan status total pemantauan.
"""

print("=== PROGRAM PEMANTAUAN BEBAN LISTRIK HARIAN ===")
print("Masukkan data beban listrik dalam kW (ketik 'selesai' untuk menganalisis data)\n")

# 1. Sebuah list untuk menyimpan data beban listrik
data_beban = []

# 2. Perulangan while untuk menginput data secara dinamis
while True:
    input_user = input("Masukkan nilai beban listrik (kW): ")

    # Kondisi untuk keluar dari perulangan input
    if input_user.lower() == 'selesai':
        break

    # Penggunaan try-except untuk validasi input angka
    try:
        beban = float(input_user)
        if beban < 0:
            print("Peringatan: Beban listrik tidak boleh bernilai negatif!")
            continue
        data_beban.append(beban) # Menyimpan data ke dalam list
    except ValueError:
        print("Input tidak valid! Masukkan angka atau ketik 'selesai'.")

# Memastikan ada data yang dianalisis
if len(data_beban) == 0:
    print("\nTidak ada data beban yang dimasukkan. Program selesai.")
else:
    print("\n" + "="*50)
    print("=== HASIL ANALISIS DAN DETEKSI ANOMALI SUHU/BEBAN ===")
    print("="*50)

    total_beban = 0
    jumlah_anomali = 0

    # 3. Perulangan for untuk memproses isi dari list data_beban
    for i, beban in enumerate(data_beban, start=1):
        total_beban += beban

        # 4. Percabangan if-elif-else untuk penentuan status beban/energi
        if beban > 500.0:
            status = "ANOMALI / OVERLOAD (Bahaya!)"
            jumlah_anomali += 1
        elif beban > 350.0:
            status = "WARNING / BEBAN TINGGI"
        else:
            status = "NORMAL"

        print(f"Data ke-{i}: {beban:.2f} kW -> Status: {status}")

    # Perhitungan statistik energi akhir menggunakan ekspresi matematika dasar
    rata_rata = total_beban / len(data_beban)

    print("-"*50)
    print(f"Total data yang diperiksa : {len(data_beban)} poin data")
    print(f"Rata-rata beban harian    : {rata_rata:.2f} kW")
    print(f"Jumlah anomali terdeteksi : {jumlah_anomali} kejadian")

    # Ringkasan akhir kondisi sistem energi
    if jumlah_anomali > 0:
        print("KESIMPULAN: Sistem dalam kondisi kritis! Segera lakukan penyeimbangan beban.")
    else:
        print("KESIMPULAN: Sistem energi berjalan stabil dan aman.")

  #PENJELASAN
  # 1. Input Data Dinamis: Program akan terus-menerus meminta pengguna untuk memasukkan data beban listrik (dalam kW) satu per satu menggunakan perulangan while True.
  # 2. Validasi Input (try-except): Setiap input diperiksa. Jika pengguna mengetik kata 'selesai', perulangan input berhenti. Jika pengguna memasukkan teks biasa (bukan angka), program menangkap error ValueError dan meminta input ulang. Jika pengguna memasukkan angka negatif, program memberikan peringatan.
  # 3. Analisis Data & Klasifikasi: Setelah input selesai, program memeriksa setiap data beban di dalam list menggunakan perulangan for dan mengklasifikasikannya ke dalam 3 kategori: ANOMALI / OVERLOAD (> 500 kW), WARNING / BEBAN TINGGI (> 350 kW), atau NORMAL.
  # 4. Output & Kesimpulan: Program menghitung rata-rata beban harian serta menghitung total kejadian anomali. Jika terdapat minimal satu anomali (> 0), sistem dinyatakan dalam kondisi kritis. Jika tidak ada, sistem dinyatakan aman.

  # 1. Inisialisasi & Input Data
    # • data_beban = [] Membuat sebuah list kosong bernama data_beban untuk menampung seluruh angka beban listrik yang dimasukkan pengguna.
    # • while True: Membuat perulangan tak terbatas agar pengguna bisa memasukkan data sebanyak apa pun secara dinamis sampai mereka memutuskan untuk berhenti.
    # • input_user = input("Masukkan nilai beban listrik (kW): ") Meminta input teks dari pengguna melalui terminal.
    # • if input_user.lower() == 'selesai': break Memeriksa apakah pengguna mengetik 'selesai' (huruf besar/kecil tidak berpengaruh karena .lower()). Jika ya, perintah break akan menghentikan perulangan while.
```
