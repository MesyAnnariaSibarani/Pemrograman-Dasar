```python
# SOAL ESAI

1. Dalam bahasa pemrograman Python, ekspresi 10 + 5 * 2 menghasilkan nilai 20, bukan 30. Hal ini terjadi karena Python menerapkan aturan prioritas operator (operator precedence) yang menentukan operator mana yang akan dieksekusi terlebih dahulu dalam sebuah ekspresi.
Di dalam ekspresi 10 + 5 * 2, terdapat dua operator: Operator penjumlahan (+) dan Operator perkalian (*)
Berdasarkan aturan prioritas di Python, operator perkalian (*) memiliki tingkat prioritas yang lebih tinggi daripada operator penjumlahan (+).
Sumber : https://www.studytrigger.com/article/python/operators-and-expressions-in-python/


2. Perbedaan mendasar antara operator relasional dan operator logika terletak pada tujuan penggunaannya.
Tujuan operator relasional adalah Membandingkan dua nilai atau variabel (angka, string, dsb) sedangkan tujuan operator logika adalah mengombinasikan atau membalikkan nilai kebenaran (boolean).

Contoh Situasi: Sebuah bank menggunakan sistem otomatis untuk menentukan apakah seorang nasabah layak mendapatkan pinjaman.

Kriterianya adalah:
1. Usia nasabah harus minimal 25 tahun.
2. Pendapatan bulanan harus lebih dari Rp7.000.000.
3. Skor kredit nasabah tidak boleh negatif (palsu/bermasalah).

Dalam situasi ini, kita menggunakan operator relasional untuk memeriksa setiap kriteria angka, dan operator logika untuk menggabungkan semua syarat tersebut.

usia = 27
pendapatan = 9000000
skor_kredit_buruk = False

# Ekspresi gabungan menggunakan operator relasional dan logika
layak_pinjam = (usia >= 25) and (pendapatan > 7000000) and (not skor_kredit_buruk)
print(layak_pinjam)


3. Salah satu contoh kasus nyata yang terkenal terjadi pada sistem perangkat lunak pesawat tempur F-22 Raptor pada tahun 2007. Saat sekelompok pesawat terbang melintasi Garis Tanggal Internasional (International Date Line), komputer di dalam pesawat mengalami kecelakaan massal (crash) pada sistem navigasi dan komunikasi. Masalah ini dipicu oleh kesalahan logika perhitungan waktu dan penanganan data koordinat, di mana baris kode tertentu mengevaluasi operasi matematika tanpa tanda kurung yang tepat, sehingga urutan eksekusi (operator precedence) menghasilkan nilai yang salah. Akibatnya, memori sistem mengalami overflow dan membuat komputer pesawat melakukan reboot berulang kali di udara.
Secara umum, kesalahan operator precedence seperti ini diperbaiki dengan menambahkan tanda kurung () secara eksplisit untuk memaksa program mengeksekusi operasi dengan urutan yang benar, bukan mengandalkan urutan bawaan bahasa pemrograman.

Jika programmer ingin menghitung rata-rata koordinat atau konversi waktu dengan rumus:
# Salah: Pembagian dikerjakan lebih dulu daripada penjumlahan
int hasil = nilaiA + nilaiB / 2;
# Benar: Tanda kurung memastikan penjumlahan dilakukan terlebih dahulu
int hasil = (nilaiA + nilaiB) / 2;

Dengan menambahkan tanda kurung, kode menjadi lebih aman, hasil perhitungan menjadi akurat, dan kesalahan fatal akibat urutan operasi bawaan compiler dapat dihindari.
```

```python
# Bukti Jawaban 3
harga_barang = 100000
diskon = 10000

# Salah: Pembagian (/) dieksekusi lebih dulu daripada pengurangan (-)
total_per_orang = harga_barang - diskon / 2
print(total_per_orang)  # Hasilnya: 95000 (Salah, karena hanya diskonnya yang dibagi 2)


harga_barang = 100000
diskon = 10000

# Benar: Tanda kurung memaksa pengurangan dihitung terlebih dahulu
total_per_orang = (harga_barang - diskon) / 2
print(total_per_orang)  # Hasilnya: 45000 (Benar)
```
