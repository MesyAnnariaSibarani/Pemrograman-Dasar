```python
# Jawaban Latihan 4

tegangan_harian = [217.5, 219.0, 220.3, 218.7, 221.1, 216.9]

# (a) cetak pembacaan pertama dan terakhir memakai indeks
print("Tegangan Pertama:", tegangan_harian[0], "Tegangan Terakhir:", tegangan_harian[5])

# (b) tambahkan satu pembacaan baru 222.0 dengan append()
tegangan_harian.append(220.0)
print(tegangan_harian)

# (c) cetak jumlah total data dengan len()
tegangan_harian = [217.5, 219.0, 220.3, 218.7, 221.1, 216.9, 220.0]
print("Jumlah data:", len(tegangan_harian))
```
