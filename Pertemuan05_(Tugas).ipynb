```python
# SOAL 1: BELANJA BUKU

jumlah_buku = 7
harga_per_buku = 18500
uang_budi = 150000

# 1. Menghitung total belanja, besar diskon, dan jumlah yang harus dibayar
total = jumlah_buku * harga_per_buku
diskon = total * 10 // 100
bayar = total - diskon

# 2. Menghitung uang kembalian
kembalian = uang_budi - bayar

# 3. Menghitung iuran per orang dan sisa pembagian (jika dipatungi 4 orang)
per_orang = bayar // 4
sisa = bayar % 4

# 4. Mengecek apakah uang Budi cukup
uang_cukup = uang_budi >= bayar

# Menampilkan output sesuai format contoh
print(f"Total        : {total}")
print(f"Diskon       : {diskon}")
print(f"Bayar        : {bayar}")
print(f"Kembalian    : {kembalian}")
print(f"Per orang    : {per_orang}")
print(f"Sisa         : {sisa}")
print(f"Uang cukup?  : {uang_cukup}")
```

```python
# SOAL 2: PLTS ATAP RUMAH

jumlah_panel = 12
daya_panel = 450  # Wp
jam_sinar = 5     # jam/hari
efisiensi = 0.8
kebutuhan_harian = 25  # kWh/hari
tarif_listrik = 1444.70

# 1. Energi yang dihasilkan per hari (kWh)
# Dibagi 1000 untuk mengubah Wh menjadi kWh
energi_harian = jumlah_panel * daya_panel * jam_sinar * efisiensi / 1000

# 2. Energi per bulan (30 hari) dan nilai rupiahnya
energi_bulanan = energi_harian * 30
nilai_listrik = energi_bulanan * tarif_listrik

# 3. Kekurangan energi per hari terhadap kebutuhan
kekurangan = kebutuhan_harian - energi_harian

# 4. Apakah PLTS sudah mencukupi kebutuhan rumah?
mencukupi = energi_harian >= kebutuhan_harian

# Menampilkan hasil output (menggunakan round() jika diperlukan untuk merapikan)
print("Energi harian (kWh) :", round(energi_harian, 1))
print("Energi bulanan (kwh):", round(energi_bulanan, 1))
print("Nilai listrik (Rp)  :", round(nilai_listrik, 1))
print("Kekurangan (kWh)    :", round(kekurangan, 1))
print("Mencukupi?          :", mencukupi)
```

```python
# SOAL 3: GENERATOR DIESEL DAN RUGI KABEL

# Diketahui data dari soal
daya_generator = 200  # kW
jam_operasi = 8       # jam/hari
konsumsi_solar = 0.25 # liter/kWh
harga_solar = 13500   # Rp/liter
kapasitas_tangki = 1000 # liter
arus = 20             # A
hambatan = 0.5        # Ohm

# 1. Energi yang dibangkitkan per hari (kWh)
energi = daya_generator * jam_operasi

# 2. Kebutuhan solar per hari dan biayanya
solar_per_hari = energi * konsumsi_solar
biaya_per_hari = solar_per_hari * harga_solar

# 3. Jumlah hari penuh tangki cukup dan sisa solar (menggunakan // dan %)
hari_cukup = kapasitas_tangki // solar_per_hari
sisa_solar = kapasitas_tangki % solar_per_hari

# 4. Rugi daya pada kabel (P = I^2 * R) menggunakan operator **
rugi_kabel = (arus ** 2) * hambatan

# 5. Evaluasi apakah rugi kabel tinggi (> 150 W dan arus >= 20 A)
rugi_tinggi = (rugi_kabel > 150) and (arus >= 20)

# Menampilkan output sesuai format contoh
print(f"Energi (kWh)   : {energi}")
print(f"Solar (liter)  : {solar_per_hari}")
print(f"Biaya (Rp)     : {biaya_per_hari}")
print(f"Tangki cukup   : {float(hari_cukup)} hari")
print(f"Sisa solar (L) : {sisa_solar}")
print(f"Rugi kabel (W) : {rugi_kabel}")
print(f"Rugi tinggi?   : {rugi_tinggi}")
```
