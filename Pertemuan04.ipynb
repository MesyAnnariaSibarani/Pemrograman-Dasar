```python
# 2. Buatlah program sederhana versi anda masing-masing dengan lingkup, fungsi input, fungsi output dan operasi artimatika

# a. LINGKUP / SCOPE VARIABEL GLOBAL
NAMA_TOKO = "Toko Mesy Jaya"

def hitung_total_belanja():
    # b. FUNGSI OUTPUT (Menampilkan menu awal)
    print(f"Selamat Datang di {NAMA_TOKO}")
    print("=" * 36)

    # c. FUNGSI INPUT (Mengambil data dari pengguna)
    # Menggunakan type casting (int dan float)
    nama_barang = input("Masukkan nama barang : ")
    harga_satuan = float(input("Masukkan harga satuan (Rp): "))
    jumlah_beli = int(input("Masukkan jumlah yang dibeli : "))

    # d. OPERASI ARITMATIKA
    # Menghitung total harga belanja kotor (perkalian)
    total_kotor = harga_satuan * jumlah_beli
    # Menghitung diskon 5% jika total kotor lebih dari Rp 300.000
    diskon = (0.05 * total_kotor) if total_kotor > 300000 else 0
    # Menghitung total bersih yang harus dibayar (pengurangan)
    total_akhir = total_kotor - diskon

    # e. FUNGSI OUTPUT (Menampilkan hasil struk belanja)
    print("\n" + "=" * 36)
    print("           STRUK BELANJA")
    print("=" * 36)
    print(f"Barang       : {nama_barang}")
    print(f"Jumlah       : {jumlah_beli}")
    print(f"Total Kotor  : Rp {total_kotor:,.2f}")
    print(f"Diskon       : Rp {diskon:,.2f}")
    print("-" * 36)
    print(f"TOTAL BAYAR  : Rp {total_akhir:,.2f}")
    print("=" * 36)
    print("Terima kasih telah berbelanja!")

# Memanggil fungsi untuk menjalankan program
hitung_total_belanja()
```
